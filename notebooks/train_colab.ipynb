# ColdStore-Vision: train yolo11s-seg on Colab

1. Runtime → Change runtime type → **T4 GPU**.
2. In Google Drive create `MyDrive/coldstore/` and upload **`coldstore_sacks_v1.zip`** (from `data/`) and **`train_yolo.py`** (from `models/`).
3. Runtime → **Run all**. Weights land in `MyDrive/coldstore/weights/`.
4. On your PC: `python -m tools.remote.fetch_weights "G:/My Drive/coldstore/weights"` (or download the folder and point at it).
